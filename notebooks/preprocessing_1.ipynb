```python
import warnings
import catppuccin
import matplotlib
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator


warnings.filterwarnings("ignore")
matplotlib.style.use(["dark_background", catppuccin.PALETTE.mocha.identifier])
%matplotlib inline
```

```python
data = pd.read_csv("../data/movie_metadata.csv")
data.columns
```

```python
plt.figure(figsize=(20, 10))
ax = sns.countplot(data=data, x="title_year")
ax.xaxis.set_major_locator(MaxNLocator(integer=True))
plt.xticks(rotation=45)
plt.legend(title="Year")
plt.title("Titles per year")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
data = data.loc[
    :,
    [
        "director_name",
        "actor_1_name",
        "actor_2_name",
        "actor_3_name",
        "genres",
        "movie_title",
        "imdb_score",
        "plot_keywords",
    ],
]
```

```python
data.shape
```

```python
data.head(5043)
```

```python
data["actor_1_name"] = data["actor_1_name"].replace(np.nan, "unknown")
data["actor_2_name"] = data["actor_2_name"].replace(np.nan, "unknown")
data["actor_3_name"] = data["actor_3_name"].replace(np.nan, "unknown")
data["director_name"] = data["director_name"].replace(np.nan, "unknown")
```

```python
data
```

```python
data["genres"] = data["genres"].str.replace("|", " ")
```

```python
data["plot_keywords"] = data["plot_keywords"].str.replace("|", " ")
data["plot_keywords"] = data["plot_keywords"].fillna("unknown")
```

```python
data
```

```python
data["movie_title"] = data["movie_title"].str.lower()
data["movie_title"] = data["movie_title"].apply(lambda x: x[:-1])
```

```python
data.to_csv("../data/data.csv", index=False)
```
